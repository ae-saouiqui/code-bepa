```python
# load the environement variables
from dotenv import load_dotenv
print(" Environement variables loaded successfully ".center(100,"*") if load_dotenv() else "Failed to load environement variables")
```

```python
# checking the file existence
import os
from pathlib import Path
root_path = os.getenv("DATA_ROOT") 
# check if this variable exists
if root_path is  None : 
    raise Exception("The varible 'DATA_ROOT' is undefined")

print("DATA_ROOT : ",root_path)

root_path = Path(root_path)

if not root_path.exists() or not root_path.is_dir() :
    raise Exception(f"{str(root_path)} is not a directory or doesn't exists")

else :
    print(f"{str(root_path)} is loaded safely")
```

```python
# reading the trainining dataset 
import polars as pl
train_path  = root_path / "train.jsonl"
if not train_path.exists() or not train_path.is_file():
    raise Exception(f"{str(train_path)} is not a file or doesn't exist")

lazy_df = pl.scan_ndjson(train_path)
```

```python
# print schema 
lazy_df.schema
```

```python
python3_df = lazy_df.filter(
    pl.col("language") == "PYTHON3"
)
```
